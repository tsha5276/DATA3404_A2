# Task 1



Find the top-10 listings in Melbourne (by id and name) by a superhost which received the most reviews in 2025, as well as the date of the latest review per each listing.


```sql
--  small
SELECT l.id, l.listing_name, COUNT(r.id) AS num_reviews, MAX(r.review_date) AS last_review_date
FROM Listings_Small l
JOIN Hosts h ON l.host_id = h.id
JOIN Reviews_Small r ON l.id = r.listing_id
JOIN Cities c ON l.city_id = c.id
WHERE c.city_name = 'Melbourne' AND h.is_superhost = 't' AND YEAR(r.review_date) = 2025
GROUP BY l.id, l.listing_name
ORDER BY num_reviews DESC
LIMIT 10;
```

```sql
--medium 
SELECT l.id, l.listing_name, COUNT(r.id) AS num_reviews, MAX(r.review_date) AS last_review_date
FROM Listings_Medium l
JOIN Hosts_medium h ON l.host_id = h.id
JOIN Reviews_Medium r ON l.id = r.listing_id
JOIN Cities c ON l.city_id = c.id
WHERE c.city_name = 'Melbourne' AND h.is_superhost = 't' AND YEAR(r.review_date) = 2025
GROUP BY l.id, l.listing_name
ORDER BY num_reviews DESC
LIMIT 10;
```

```sql
-- large
SELECT l.id, l.listing_name, COUNT(r.id) AS num_reviews, MAX(r.review_date) AS last_review_date
FROM Listings_Large l
JOIN Hosts h ON l.host_id = h.id
JOIN Reviews_Large r ON l.id = r.listing_id
JOIN Cities c ON l.city_id = c.id
WHERE c.city_name = 'Melbourne' AND h.is_superhost = 't' AND YEAR(r.review_date) = 2025
GROUP BY l.id, l.listing_name
ORDER BY num_reviews DESC
LIMIT 10;
```

```sql
EXPLAIN SELECT l.id, l.listing_name, COUNT(r.id) AS num_reviews, MAX(r.review_date) AS last_review_date
FROM Listings_Large l
JOIN Hosts h ON l.host_id = h.id
JOIN Reviews_Large r ON l.id = r.listing_id
JOIN Cities c ON l.city_id = c.id
WHERE c.city_name = 'Melbourne' AND h.is_superhost = 't' AND YEAR(r.review_date) = 2025
GROUP BY l.id, l.listing_name
ORDER BY num_reviews DESC
LIMIT 10;
```

```sql
-- small listing
SELECT c.city_name, n.nhood_name, COUNT(DISTINCT l.id) AS num_listings, COUNT(r.id) AS num_positive_reviews, ROUND(AVG(l.rating), 2) AS avg_rating, MAX(r.review_date) AS most_recent_review
FROM Listings_Small l
JOIN Hosts h ON l.host_id = h.id
JOIN Neighbourhoods n ON l.neighbourhood = n.id
JOIN Cities c ON l.city_id = c.id
JOIN Reviews_Small r ON l.id = r.listing_id
WHERE c.country = 'Australia' AND h.is_verified = 't' AND l.amenities LIKE '%Wifi%' AND (LOWER(r.comments) LIKE '%great place%'  OR LOWER(r.comments) LIKE '%clean place%')
GROUP BY c.city_name, n.nhood_name
HAVING COUNT(r.id) >= 650
ORDER BY num_positive_reviews DESC
LIMIT 10;
```

```sql
-- medium listing
SELECT c.city_name, n.nhood_name, COUNT(DISTINCT l.id) AS num_listings, COUNT(r.id) AS num_positive_reviews, ROUND(AVG(l.rating), 2) AS avg_rating, MAX(r.review_date) AS most_recent_review
FROM Listings_Medium l
JOIN Hosts h ON l.host_id = h.id
JOIN Neighbourhoods n ON l.neighbourhood = n.id
JOIN Cities c ON l.city_id = c.id
JOIN Reviews_Medium r ON l.id = r.listing_id
WHERE c.country = 'Australia' AND h.is_verified = 't' AND l.amenities LIKE '%Wifi%' AND (LOWER(r.comments) LIKE '%great place%' OR LOWER(r.comments) LIKE '%clean place%')
GROUP BY c.city_name, n.nhood_name
HAVING COUNT(r.id) >= 650
ORDER BY num_positive_reviews DESC
LIMIT 10;
```

```sql
-- large listing
SELECT c.city_name, n.nhood_name, COUNT(DISTINCT l.id) AS num_listings, COUNT(r.id) AS num_positive_reviews, ROUND(AVG(l.rating), 2) AS avg_rating, MAX(r.review_date) AS most_recent_review
FROM Listings_Large l
JOIN Hosts h ON l.host_id = h.id
JOIN Neighbourhoods n ON l.neighbourhood = n.id
JOIN Cities c ON l.city_id = c.id
JOIN Reviews_Large r ON l.id = r.listing_id
WHERE c.country = 'Australia' AND h.is_verified = 't' AND l.amenities LIKE '%Wifi%' AND (LOWER(r.comments) LIKE '%great place%' OR LOWER(r.comments) LIKE '%clean place%')
GROUP BY c.city_name, n.nhood_name
HAVING COUNT(r.id) >= 650
ORDER BY num_positive_reviews DESC
LIMIT 10;
```

```sql
%python
# load dfs

from pyspark.sql import functions as F

listings = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("nullValue", "NULL") \
    .option("dateFormat", "yyyy-MM-dd") \
    .option("samplingRatio", "0.01") \
    .load("/Volumes/workspace/data3404/airbnb/airbnb_listings-small.csv")

reviews = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("nullValue", "NULL") \
    .option("dateFormat", "yyyy-MM-dd") \
    .option("samplingRatio", "0.01") \
    .load("/Volumes/workspace/data3404/airbnb/airbnb_reviews-small.csv")

cities = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("nullValue", "NULL") \
    .load("/Volumes/workspace/data3404/airbnb/airbnb_cities.csv")

neighbourhoods = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("nullValue", "NULL") \
    .load("/Volumes/workspace/data3404/airbnb/airbnb_neighbourhoods.csv")
```

```sql
%python

city_name = "Melbourne"

city_df = cities.filter(F.col("city_name") == city_name)
city_id = city_df.first()["id"]

# remove hotel/hostel room types
listings_filtered = listings.filter(
    (F.col("city_id") == city_id) &
    (~F.lower(F.col("property_type")).like("%hotel%")) &
    (~F.lower(F.col("property_type")).like("%hostel%"))
)

listings_filtered.show(5)
```

```sql
%python
# Join reviews with listings to get minimum_nights
reviews_listings = reviews.join(
    listings_filtered,
    reviews.listing_id == listings_filtered.id,
    "inner"
)

# Calculate booking duration:
# max(3, minimum_nights) but capped at 21
booking_duration = reviews_listings.withColumn(
    "duration",
    F.least(F.lit(21), F.greatest(F.lit(3), F.col("minimum_nights")))
)

booking_duration.select("listing_id", "minimum_nights", "duration").show(5)
```

```sql
%python
# Filter reviews to 2025 only
reviews_2025 = booking_duration.filter(
    (F.col("review_date") >= '2025-01-01') & 
    (F.col("review_date") < '2026-01-01')
)

# Calculate occupied days per listing
# Formula: (num_reviews / 0.67) * duration
occupied_days = reviews_2025.groupBy("listing_id", "minimum_nights") \
    .agg(
        F.count("*").alias("num_reviews"),
        F.first("duration").alias("duration"),
        F.first("neighbourhood").alias("neighbourhood"),
        F.first("rating").alias("rating"),
        F.first("property_type").alias("property_type"),
        F.first("listing_name").alias("listing_name")
    ) \
    .withColumn("occupied_days", (F.col("num_reviews") / 0.67) * F.col("duration"))

occupied_days.show(5)
```

```sql
%python
# Calculate occupancy rate as a percentage
occupancy = occupied_days.withColumn(
    "occupancy_rate",
    F.least(F.lit(1.0), F.col("occupied_days") / F.lit(365.0))
)

occupancy.select("listing_id", "listing_name", "occupied_days", "occupancy_rate").show(5)
```
